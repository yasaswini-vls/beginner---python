```python
x="*"
count = 1
while count <= 5:
  print(x)
  x=x+"*"
  count = count+1
```

```python
x="1"
count = 1
i=1
while count <= 5:
  print(x)
  x=x+str(i+1)
  i=i+1
  count=count+1
```
