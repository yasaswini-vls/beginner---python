```python
import random
z=random.randint(1,20)
user=int(input("guess the number between 1 to 20"))
while user != z:
  if user > z:
    print("too high")
    user=int(input("guess the number between 1 to 20"))
  if user < z:
    print("too low")
    user=int(input("guess the number between 1 to 20"))
if user == z:
  print("youve got it")
```
