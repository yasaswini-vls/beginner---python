```python
print("Game")
import random
r=int(input("How many rounds"))
p=(input("type roll to start"))
i=1
while i<=r and p=='roll':
  x=random.randint(1,6)
  y=random.randint(1,6)
  print("you got", str(x))
  print("computer got", str(y))
  if x>y:
    print("you win")
  elif x==y:
    print("tie")
  else:
    print("you win")
  print("you have",str(3-i),"attempts left")
  p=(input("type roll to start"))
  i=i+1

```
