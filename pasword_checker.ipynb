```python
user=input("Enter your password")
has_num = False
while len(user)<8:
  print("weak password min 8 characters required")
  user=input("Enter your password")
if user == "password" or user == "abcdefg" or user == "12345678":
  print("weak")
  user=input("Enter your password")
spch = "!@#$%^&*"
num = '1,2,3,4,5,6,7,8,9,0'
for ch in user:
  if ch in spch or ch in num:
    has_num = True
if has_num:
  print("strong")
  print("password updated")
```
