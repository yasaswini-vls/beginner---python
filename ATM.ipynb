```python
print("=======ATM========")
print("Welcome to this ATM, please choose one of the following options")
print("1-Check Balance \n 2-Deposit \n 3-Withdrawal \n 4-Exit")
balance = 1000
choice = int(input("Enter your option"))
while choice != 4:
  if choice == 1:
    print("balance is",str(balance))
  elif choice == 2:
    deposit = int(input("Enter amount you wish to deposit"))
    balance = balance + deposit
    print("balance is", str(balance))
  elif choice == 3:
    withdraw = int(input("Enter amount you wish to withdraw"))
    if withdraw > balance:
      print("insufficient funds")
    else:
      balance = balance-withdraw
      print("balance is",str(balance))
  else:
      print("thank you")
  choice = int(input("Enter your option"))
print("thanks for using ATM")
```
