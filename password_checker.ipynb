```python
print("LOGIN")
tr_loginid = "Yasaswini@16"
tr_pwd = "@Apple456"
user_loginid = input("Enter loginid")
user_pwd = input("Enter password")
attempt = 1

while tr_loginid != user_loginid and tr_pwd != user_pwd and attempt <=3:
    print("try again")
    print("you have ",3-attempt,"attempts left")
    user_loginid = input("Enter loginid")
    user_pwd = input("Enter password")
    attempt = attempt+1
if tr_loginid == user_loginid and tr_pwd == user_pwd:
  print("SUCCESS")
else:
  print('LOCKED')
```
