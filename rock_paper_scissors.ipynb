```python
import random
list = ['Rock','Paper',"Scissor"]
print("lets play rock paper scissors")
user = str(input("lets play"))
z=random.choice(list)
print(z)
print("I picked", z)
if z == 'Rock' and user == 'Scissor' or z == 'Scissor' and user == 'Paper' or z == 'Paper' and user == 'Rock':
    print("you lose")
if z == 'Rock' and user == 'Paper' or z == 'Scissor' and user == 'Rock' or z == 'Paper' and user == 'Scissor':
    print("you win")
elif z == user:
    print('its a tie')
    user = str(input("lets play"))
print("wanna play again?")


```
